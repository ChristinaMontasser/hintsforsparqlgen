```python
from pathlib import Path


def local_name(uri_or_prefixed_name: str) -> str:
    """
    Extract local name from:
    - https://w3id.org/hmontology/ClConcentrationMean
    - hyobs:ClConcentrationMean
    - ClConcentrationMean
    """
    value = uri_or_prefixed_name.strip()
    value = value.strip("<>")
    value = value.rstrip()

    if "#" in value:
        value = value.split("#")[-1]
    if "/" in value:
        value = value.split("/")[-1]
    if ":" in value:
        value = value.split(":")[-1]

    return value


def update_schema_observed_properties(
    schema_file: str | Path,
    observed_properties: list[str],
    graph_name: str,
    output_dir: str | Path | None = None,
) -> Path:
    schema_file = Path(schema_file)

    if output_dir is None:
        output_dir = schema_file.parent
    else:
        output_dir = Path(output_dir)
        output_dir.mkdir(parents=True, exist_ok=True)

    output_file = output_dir / f"{schema_file.stem}_{graph_name.replace('/', '_')}.txt"

    allowed_names = sorted({local_name(p) for p in observed_properties})

    kept_lines = []
    removed_observed_property_lines = []

    # Step 1: read original schema and remove ALL existing observedProperty lines
    with open(schema_file, "r", encoding="utf-8") as f:
        for line in f:
            if line.strip().startswith("sosa:observedProperty:"):
                removed_observed_property_lines.append(line)
            else:
                kept_lines.append(line)

    # Step 2: append only graph-specific observedProperty lines
    new_observed_property_lines = [
        f"sosa:observedProperty: sosa:Observation -> hyobs:{name}\n"
        for name in allowed_names
    ]

    # Optional separator for readability
    if kept_lines and not kept_lines[-1].endswith("\n"):
        kept_lines[-1] += "\n"

    final_lines = (
        kept_lines
        + ["\n# Graph-specific observed properties\n"]
        + new_observed_property_lines
    )

    # Step 3: save new graph-specific schema file
    with open(output_file, "w", encoding="utf-8") as f:
        f.writelines(final_lines)

    print(f"[{graph_name}] saved schema: {output_file}")
    print(f"[{graph_name}] removed old observedProperty rows: {len(removed_observed_property_lines)}")
    print(f"[{graph_name}] added new observedProperty rows: {len(new_observed_property_lines)}")

    return output_file
```

```python
schema_file = (
    "helper/hydroOntology/schema_profiling/"
    "260609_1530_predicate_subject_object_namespaces_full.txt"
)

graph_observed_properties = {
    "quadica": quadica_observedProperties,
    "lamah_ce": lamah_ce_observedProperties,
    "camels_de": camels_gb_observedProperties,
}

for graph_name, observed_properties in graph_observed_properties.items():
    update_schema_observed_properties(
        schema_file=schema_file,
        observed_properties=observed_properties,
        graph_name=graph_name,
    )

#Done 
```

```python
from pathlib import Path
import re


RELATION_RE = re.compile(
    r"""
    (?P<predicate>[A-Za-z_][\w.-]*:[A-Za-z_][\w.-]*)  # e.g. sosa:observedProperty
    :\s*
    (?P<subject>[A-Za-z_?][\w.:-]*)                   # e.g. sosa:Observation or ?x
    \s*->\s*
    (?P<object>[A-Za-z_?][\w.:-]*)                    # e.g. hyobs:air_temperature or ?x
    """,
    re.VERBOSE,
)


def extract_relations(text: str) -> list[tuple[str, str, str]]:
    """
    Extract relations from one long schema-profile text.

    Returns:
    [(predicate, subject, object), ...]
    """
    return [
        (
            match.group("predicate"),
            match.group("subject"),
            match.group("object"),
        )
        for match in RELATION_RE.finditer(text)
    ]


def local_name(value: str) -> str:
    value = value.strip().strip("<>").rstrip(" .,;\n\t")

    if "#" in value:
        value = value.split("#")[-1]
    if "/" in value:
        value = value.split("/")[-1]
    if ":" in value:
        value = value.split(":")[-1]

    return value


def update_schema_for_graph_from_one_line_text(
    schema_file: str | Path,
    observed_properties: list[str],
    has_properties: list[str],
    graph_name: str,
    output_dir: str | Path | None = None,
) -> Path:
    schema_file = Path(schema_file)

    if output_dir is None:
        output_dir = schema_file.parent
    else:
        output_dir = Path(output_dir)
        output_dir.mkdir(parents=True, exist_ok=True)

    output_file = output_dir / f"{schema_file.stem}_{graph_name.replace('/', '_')}.txt"

    text = schema_file.read_text(encoding="utf-8")
    relations = extract_relations(text)
    print(relations)
    allowed_observed = sorted({local_name(p) for p in observed_properties})
    allowed_has = sorted({local_name(p) for p in has_properties})

    final_relations = []
    removed_observed_count = 0
    removed_has_count = 0

    for predicate, subject, obj in relations:
        if predicate == "sosa:observedProperty":
            removed_observed_count += 1
            continue

        if predicate.startswith("hyobs:has") and subject == "envthes:30212":
            removed_has_count += 1
            continue

        final_relations.append((predicate, subject, obj))

    for name in allowed_observed:
        final_relations.append(
            ("sosa:observedProperty", "sosa:Observation", f"hyobs:{name}")
        )

    for name in allowed_has:
        final_relations.append(
            (f"hyobs:{name}", "envthes:30212", "?x")
        )

    output_text = " ".join(
        f"{predicate}: {subject} -> {obj}"
        for predicate, subject, obj in final_relations
    )

    output_file.write_text(output_text + " ", encoding="utf-8")

    print(f"[{graph_name}] saved: {output_file}")
    print(f"[{graph_name}] parsed relations: {len(relations)}")
    print(f"[{graph_name}] removed observedProperty relations: {removed_observed_count}")
    print(f"[{graph_name}] added observedProperty relations: {len(allowed_observed)}")
    print(f"[{graph_name}] removed hyobs:has* catchment relations: {removed_has_count}")
    print(f"[{graph_name}] added hyobs:has* catchment relations: {len(allowed_has)}")

    return output_file



schema_file = (
    "helper/hydroOntology/schema_profiling/"
    "260609_1530_predicate_subject_object_namespaces_full.txt"
)

update_schema_for_graph_from_one_line_text(
    schema_file=schema_file,
    observed_properties=quadica_observed_property_names,
    has_properties=quadica_has_properties,
    graph_name=GRAPH_PATH,
)
```

```python
from pathlib import Path
import re

api = ENDPOINT

def extract_query_properties(response: dict) -> set[str]:
    """
    Extract property URIs from validate_query(...) response.
    Expected response shape:
    {
      "success": True,
      "data": {
        "results": {
          "bindings": [...]
        }
      }
    }
    """
    if not response.get("success"):
        raise RuntimeError(response.get("error"))

    bindings = response["data"]["results"]["bindings"]

    return {
        binding["property"]["value"]
        for binding in bindings
        if "property" in binding
    }


def local_name(value: str) -> str:
    """
    Converts:
    https://w3id.org/hmontology/hasCatchmentArea -> hasCatchmentArea
    http://www.w3.org/ns/sosa/observedProperty -> observedProperty
    hyobs:hasCatchmentArea -> hasCatchmentArea
    sosa:observedProperty -> observedProperty
    """
    value = value.strip().strip("<>").rstrip(" .,\n\t")

    if "#" in value:
        value = value.split("#")[-1]

    if "/" in value:
        value = value.split("/")[-1]

    if ":" in value:
        value = value.split(":")[-1]

    return value


def extract_schema_properties_from_text_file(schema_file: str | Path) -> set[str]:
    """
    Extract predicates from your schema profile file.

    Works even if the schema file is one long text line.

    Expected relation shape:
    predicate: subject -> object

    Examples:
    hyobs:hasCatchmentArea: envthes:30212 -> ?x
    sosa:observedProperty: sosa:Observation -> hyobs:30106
    qudt:numericValue: qudt:QuantityValue -> ?x
    """

    schema_file = Path(schema_file)
    text = schema_file.read_text(encoding="utf-8")

    # Match predicate part before ": subject -> object"
    # Handles predicates like hyobs:hasCatchmentArea, sosa:observedProperty, geo:hasGeometry
    pattern = re.compile(
        r"([A-Za-z_][\w.-]*:[A-Za-z_][\w.-]*)\s*:\s*[^>]+?\s*->\s*[^ ]+"
    )

    return {
        match.group(1)
        for match in pattern.finditer(text)
    }


def property_uri_to_prefixed(uri: str) -> str:
    for namespace, prefix in PREFIX_MAP.items():
        if uri.startswith(namespace):
            return f"{prefix}:{uri[len(namespace):]}"
    return uri

def check_schema_file_against_graph_properties(
    graph_name: str,
    schema_file: str | Path,
    api,
):
    query = f"""
SELECT DISTINCT ?property
WHERE {{
  GRAPH <{graph_name}> {{
    ?s ?property ?o .
  }}
}}
ORDER BY ?property
"""

    response = validate_query(query, api)

    query_property_uris = extract_query_properties(response)

    # Convert query result URIs to prefixed names
    query_properties_prefixed = {
        property_uri_to_prefixed(uri)
        for uri in query_property_uris
    }

    # Read properties from your graph-specific schema file
    schema_properties = extract_schema_properties_from_text_file(schema_file) #triple store (- observedProperties, ---)

    # In schema file but not actually present in the target graph
    missing_from_graph = sorted(schema_properties - query_properties_prefixed)

    # Present in graph but missing from schema file
    missing_from_schema_file = sorted(query_properties_prefixed - schema_properties)

    print("=" * 80)
    print(f"Graph: {graph_name}")
    print(f"Schema file: {schema_file}")
    print(f"Properties in graph query: {len(query_properties_prefixed)}")
    print(f"Properties in schema file: {len(schema_properties)}")

    print("\nProperties in schema file but NOT found in graph:")
    if missing_from_graph:
        for p in missing_from_graph:
            print(f"  - {p}")
    else:
        print("  None")

    print("\nProperties in graph but NOT found in schema file:")
    if missing_from_schema_file:
        for p in missing_from_schema_file:
            print(f"  - {p}")
    else:
        print("  None")

    return {
        "graph_name": graph_name,
        "graph_properties": query_properties_prefixed,
        "schema_properties": schema_properties,
        "missing_from_graph": missing_from_graph,
        "missing_from_schema_file": missing_from_schema_file,
    }


from pathlib import Path
import re


SCHEMA_RELATION_PATTERN = re.compile(
    r"(?P<predicate>[A-Za-z_][\w.-]*:[A-Za-z_][\w.-]*)\s*:\s*"
    r"(?P<subject>[^\s]+)\s*->\s*"
    r"(?P<object>[^\s]+)"
)


def extract_schema_relations(schema_text: str) -> list[dict]:
    """
    Extract schema relations from one-line or multiline schema text.

    Example relation:
    hyobs:hasCatchmentArea: envthes:30212 -> qudt:QuantityValue
    """
    relations = []

    for match in SCHEMA_RELATION_PATTERN.finditer(schema_text):
        relations.append(
            {
                "predicate": match.group("predicate"),
                "subject": match.group("subject"),
                "object": match.group("object"),
                "text": (
                    f"{match.group('predicate')}: "
                    f"{match.group('subject')} -> "
                    f"{match.group('object')}"
                ),
            }
        )

    return relations


def create_schema_without_missing_properties(
    schema_file: str | Path,
    missing_properties: list[str] | set[str],
    output_file: str | Path | None = None,
    compare_by_local_name: bool = False,
) -> Path:
    """
    Create a new schema file after removing relations whose predicate is missing
    from the actual graph.

    Parameters
    ----------
    schema_file:
        Existing graph-specific schema file.

    missing_properties:
        Properties printed under:
        'Properties in schema file but NOT found in graph'

    output_file:
        Optional custom output path. If None, saves as:
        <original_stem>_graph_validated.txt

    compare_by_local_name:
        If True, compare only local names.
        Example: hyobs:hasCatchmentArea and some/full/uri/hasCatchmentArea
        are treated as the same local property.
    """
    schema_file = Path(schema_file)

    if output_file is None:
        output_file = schema_file.with_name(
            f"{schema_file.stem}_graph_validated{schema_file.suffix}"
        )
    else:
        output_file = Path(output_file)

    text = schema_file.read_text(encoding="utf-8")
    relations = extract_schema_relations(text)

    if compare_by_local_name:
        missing_set = {local_name(p) for p in missing_properties}
    else:
        missing_set = set(missing_properties)

    kept_relations = []
    removed_relations = []

    for relation in relations:
        predicate = relation["predicate"]

        key = local_name(predicate) if compare_by_local_name else predicate

        if key in missing_set:
            removed_relations.append(relation)
        else:
            kept_relations.append(relation)

    output_text = "\n".join(r["text"] for r in kept_relations) + "\n"
    output_file.write_text(output_text, encoding="utf-8")

    print("=" * 80)
    print(f"Original schema file: {schema_file}")
    print(f"New cleaned schema file: {output_file}")
    print(f"Original relation count: {len(relations)}")
    print(f"Removed relation count: {len(removed_relations)}")
    print(f"Kept relation count: {len(kept_relations)}")

    print("\nRemoved properties:")
    for r in removed_relations:
        print(f"  - {r['predicate']}")

    return output_file


def check_and_create_clean_schema_file(
    graph_name: str,
    schema_file: str | Path,
    api,
    output_file: str | Path | None = None,
    compare_by_local_name: bool = False,
):
    result = check_schema_file_against_graph_properties(
        graph_name=graph_name,
        schema_file=schema_file,
        api=api,
    )

    cleaned_file = create_schema_without_missing_properties(
        schema_file=schema_file,
        missing_properties=result["missing_from_graph"],
        output_file=output_file,
        compare_by_local_name=compare_by_local_name,
    )

    result["cleaned_schema_file"] = cleaned_file

    return result
api = ENDPOINT

```
